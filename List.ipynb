```python
# Creating a list
my_list = [10, 20, 30, 40, 20]
print("List:", my_list)
```

```python
# append()
numbers = [10, 20, 30]
numbers.append(40)
print("After append:", numbers)
```

```python
# extend()
numbers = [1, 2, 3]
numbers.extend([4, 5, 6])
print("After extend:", numbers)
```

```python
# insert()
numbers = [1, 3, 4]
numbers.insert(1, 2)  # inserting 2 at index 1
print("After insert:", numbers)
```

```python
# remove()
numbers = [1, 2, 3, 2]
numbers.remove(2)  # removes the first 2
print("After remove:", numbers)
```

```python
# pop()
numbers = [10, 20, 30]
popped_value = numbers.pop()  # removes 30
print("Popped value:", popped_value)
print("After pop:", numbers)
```

```python
# clear()
numbers = [10, 20, 30]
numbers.clear()
print("After clear:", numbers)
```

```python
# index()
numbers = [10, 20, 30, 40]
index_pos = numbers.index(30)
print("Index of 30:", index_pos)
```

```python
# count()
numbers = [1, 2, 3, 2, 4, 2]
count_2 = numbers.count(2)
print("Count of 2:", count_2)
```

```python
# reverse()=
numbers = [1, 2, 3, 4]
numbers.reverse()
print("After reverse:", numbers)
```

```python
# sort()
numbers = [3, 1, 4, 2]
numbers.sort()
print("After sort:", numbers)
```

```python
# copy()
numbers = [1, 2, 3]
copy_numbers = numbers.copy()
print("Copied list:", copy_numbers)
```

```python
# len()
numbers = [1, 2, 3, 4]
print("Length of list:", len(numbers))
```

```python
# sum()
numbers = [1, 2, 3, 4]
total = sum(numbers)
print("Sum of list:", total)
```

```python
# max()
# min()
numbers = [5, 8, 2, 10, 3]
print("Max value:", max(numbers))
print("Min value:", min(numbers))
```

Append the number 100 to the end of the list

```python
#create list
numbers = [10, 20, 30, 40]
# Append 100 to the list
numbers.append(100)
print("Updated list after append:", numbers)
```

Insert the number 50 at the 3rd position (index 2).

```python
numbers = [10, 20, 30, 40]
# Insert 50 at index 2 (3rd position)
numbers.insert(2, 50)
print("Updated list after insert:", numbers)
```

Remove the first occurrence of the number 20

```python
numbers = [10, 20, 30, 40, 20]
# Remove the first occurrence of 20
numbers.remove(20)
print("Updated list after removing 20:", numbers)
```

Count how many times 10 appears in the list.

```python
numbers = [10, 20, 30, 10, 40, 10]
# Count how many times 10 appears
count_10 = numbers.count(10)
print("Number of times 10 appears:", count_10)
```

Reverse a List

```python
numbers = [10, 20, 30, 40, 50]
# Reverse the list
numbers.reverse()
print("List after reverse:", numbers)
```

Sort the list in ascending order

```python
numbers = [50, 20, 40, 10, 30]
# Sort the list in ascending order
numbers.sort()
print("List after sort:", numbers)
```

Find the maximum and minimum values in the list

```python
numbers = [50, 20, 40, 10, 30]
# Find maximum and minimum values
maximum_value = max(numbers)
minimum_value = min(numbers)
print("Maximum value:", maximum_value)
print("Minimum value:", minimum_value)
```

Sum of all elements in the list

```python
numbers = [50, 20, 40, 10, 30]
# Calculate the sum of all elements
total_sum = sum(numbers)
print("Sum of list:", total_sum)
```
